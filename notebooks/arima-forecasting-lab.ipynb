# ARIMA / SARIMA forecasting lab
Synthetic monthly series; all model fitting is retrospective and the example is for learning, not a forecast guarantee.

```python
import sys
sys.path.insert(0, '../implementations/python/arima-forecasting')
from arima_forecasting import fit_arima
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng=np.random.default_rng(11)
t=np.arange(144)
y=0.025*t + 2*np.sin(2*np.pi*t/12) + rng.normal(0,0.3,len(t))
train=y[:120]
fit=fit_arima(train,order=(1,1,1),steps=24,seasonal_order=(1,0,0,12))
fit['aic'],fit['bic'],fit['converged']
```

```python
plt.figure(figsize=(10,4))
plt.plot(np.arange(len(y)),y,label='observed')
future=np.arange(120,144)
f=np.array(fit['forecast']); ci=np.array(fit['confidence_interval'])
plt.plot(future,f,label='forecast')
plt.fill_between(future,ci[:,0],ci[:,1],alpha=0.2,label='model interval')
plt.axvline(119,linestyle='--'); plt.legend(); plt.grid(True)
```

## Interpretation
Compare the forecast with a seasonal-naive baseline. Inspect residual autocorrelation before trusting the model. A model-based interval is not automatically calibrated, and the example does not encode spatial dependence.
