# GP-UCB spatial survey selection laboratory
Choose the next survey site from a candidate grid using a Gaussian-process upper-confidence-bound acquisition. Hyperparameters are fixed to keep the selection rule visible.

```python
import sys, numpy as np
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'implementations/python/adaptive-sampling'))
from bayesian_optimization import gp_ucb_select
observed=np.array([[0,0],[2,1],[4,0]],dtype=float)
values=np.array([0.2,1.1,0.3])
axis=np.linspace(-1,5,40);xx,yy=np.meshgrid(axis,axis);candidates=np.column_stack((xx.ravel(),yy.ravel()))
result=gp_ucb_select(observed,values,candidates,length_scale=1.4,signal_variance=1,noise_variance=.05,beta=2)
selected=candidates[result['selected_index']]
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,2,figsize=(11,4))
for a,z,title in zip(ax,[result['posterior_mean'],result['ucb']],['Posterior mean','UCB acquisition']):
    im=a.scatter(candidates[:,0],candidates[:,1],c=z,s=14);a.scatter(observed[:,0],observed[:,1],c='black',marker='x',label='observed');a.scatter(*selected,c='red',marker='*',s=180,label='selected');a.set_title(title);a.legend();fig.colorbar(im,ax=a)
plt.tight_layout()
```

## Interpretation
Increase β to favour uncertainty, or reduce it to favour the posterior mean. The isotropic RBF model assumes stationary similarity in Euclidean coordinates. Barriers, anisotropy, heterogeneous sampling costs, nonstationarity and design-based inclusion probabilities need explicit extensions before operational survey use.
