# Spatial cross-validation splitters laboratory
Compare grid-block, buffered leave-one-out and deterministic cluster-based folds. These designs answer different prediction questions and should not be selected by their score.

```python
import sys, numpy as np
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'implementations/python/spatial-cross-validation'))
from spatial_cross_validation import spatial_block_folds,buffered_leave_one_out_folds,spatial_kfold_by_clustering
rng=np.random.default_rng(8)
xy=np.vstack([rng.normal([0,0],.8,(20,2)),rng.normal([5,0],.8,(20,2)),rng.normal([2.5,4],.8,(20,2))])
```

```python
cluster_folds=spatial_kfold_by_clustering(xy,3)
block_folds=spatial_block_folds(xy,3,block_size=2.0,seed=4)
buffered_loo=buffered_leave_one_out_folds(xy,buffer_distance=1.0)
[(len(f.train_indices),len(f.test_indices)) for f in cluster_folds],[(len(f.train_indices),len(f.test_indices)) for f in block_folds]
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,2,figsize=(10,4))
for k,fold in enumerate(cluster_folds): ax[0].scatter(xy[fold.test_indices,0],xy[fold.test_indices,1],label=f'cluster fold {k+1}')
for k,fold in enumerate(block_folds): ax[1].scatter(xy[fold.test_indices,0],xy[fold.test_indices,1],label=f'block fold {k+1}')
for a,title in zip(ax,['Cluster-based folds','Grid-block folds']):a.set_title(title);a.axis('equal');a.legend();a.grid(True)
plt.tight_layout()
```

## Interpretation
Cluster-based folds group by coordinate proximity but are not guaranteed to be contiguous or equal-sized. Grid-block folds depend on block size and origin. Buffered LOO excludes nearby training observations around each held-out point. None is NNDM/kNNDM, which match prediction-to-training distance distributions. Inspect maps, distances and fold composition before reporting performance.
