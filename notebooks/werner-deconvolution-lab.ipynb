# Werner 2D thin-dike deconvolution lab
Generate an idealized profile, solve overlapping seven-point windows, and inspect the recovered top position and depth.

```python
import sys
sys.path.insert(0, '../implementations/python/potential-field')
import numpy as np
import matplotlib.pyplot as plt
from werner_deconvolution import werner_thin_dike_profile, werner_deconvolution_windows
x = np.linspace(-10, 10, 81)
truth = dict(x0=1.25, depth=2.8, amplitude_a=8.0, amplitude_b=-2.5)
anomaly = werner_thin_dike_profile(x, **truth)
result = werner_deconvolution_windows(x, anomaly, window_size=7, step=2)
print('Accepted windows:', len(result['solutions']), 'rejected:', len(result['rejected_windows']))
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, anomaly, label='Synthetic anomaly')
ax.axvline(truth['x0'], linestyle='--', label='True source x0')
ax.set_xlabel('Profile distance'); ax.set_ylabel('Anomaly'); ax.legend(); plt.show()
estimates = np.array([[s['x0'], s['depth']] for s in result['solutions']])
fig, ax = plt.subplots(figsize=(7, 3))
ax.scatter(estimates[:, 0], estimates[:, 1], s=18)
ax.axvline(truth['x0'], linestyle='--'); ax.axhline(truth['depth'], linestyle='--')
ax.set_xlabel('Estimated x0'); ax.set_ylabel('Estimated top depth')
ax.set_title('Window solutions: compare with the known synthetic source'); plt.show()
```

## Interpretation
The ideal synthetic profile is intentionally matched to the single thin-dike equation. Real data may contain regional fields, overlapping sources, noise, remanence and departures from the 2D assumptions. This lab does not perform multi-source clustering or nonlinear refinement.
