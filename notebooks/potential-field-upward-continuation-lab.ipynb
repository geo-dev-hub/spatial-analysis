# Potential-field upward continuation lab
Explore the spectral operator that continues a regular potential-field grid upward by height h. Each Fourier mode is attenuated by exp(-k h); the mean (zero wavenumber) is unchanged.

```python
import sys
sys.path.insert(0, '../implementations/python/potential-field')
import numpy as np
import matplotlib.pyplot as plt
from spectral_derivatives import upward_continuation
n=128; dx=1.0; axis=np.arange(n)*dx
x,y=np.meshgrid(axis,axis)
field=np.cos(2*np.pi*x/32)+0.5*np.cos(2*np.pi*y/12)+0.2*np.cos(2*np.pi*(x+y)/6)
heights=[0,2,5,10]
continued=[upward_continuation(field,dx,dx,h) for h in heights]
```

```python
fig,axes=plt.subplots(1,len(heights),figsize=(15,4),constrained_layout=True)
for ax,z,h in zip(axes,continued,heights):
    im=ax.imshow(z,origin='lower');ax.set_title(f'height = {h}');fig.colorbar(im,ax=ax)
plt.show()
```

## Interpretation
Short-wavelength features are attenuated more rapidly than long-wavelength features. This is a smoothing continuation, not downward continuation. The FFT assumes periodic boundaries: real survey edges can produce ringing, so inspect padding and tapering choices before interpreting a real grid.
