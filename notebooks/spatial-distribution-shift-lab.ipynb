# Spatial distribution-shift laboratory
Compare two synthetic feature distributions using unbiased MMD, energy distance and an exchangeability-based permutation test.

```python
import sys, numpy as np
sys.path.insert(0,'../implementations/python/distribution-shift')
from distribution_shift import mmd_squared,energy_distance,mmd_permutation_test
rng=np.random.default_rng(5)
x=rng.normal(0,1,size=(80,2))
y=rng.normal([.8,0],1,size=(80,2))
mmd_squared(x,y,bandwidth=1),energy_distance(x,y)
```

```python
test=mmd_permutation_test(x,y,bandwidth=1,n_permutations=199,random_state=12)
test['statistic'],test['p_value']
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,5));plt.scatter(x[:,0],x[:,1],label='X',alpha=.6);plt.scatter(y[:,0],y[:,1],label='Y',alpha=.6);plt.axis('equal');plt.legend();plt.title('Synthetic feature distributions')
```

## Spatial caveat
The permutation test assumes exchangeable independent rows. Spatial autocorrelation violates this assumption. A spatial application needs a block/cluster-aware or model-based null design; a small p-value from naive row permutation is not valid evidence of spatial distribution shift.
