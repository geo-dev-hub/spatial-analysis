# Ear clipping: watch a concave polygon become triangles

This lab uses the tested teaching implementation. The input is a simple polygon ring; holes and self-intersections are outside its contract.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
root = Path.cwd()
while root != root.parent and not (root / 'implementations/python/ear-clipping').is_dir():
    root = root.parent
package = root / 'implementations/python/ear-clipping'
if not package.is_dir(): raise RuntimeError('Could not locate implementations/python/ear-clipping')
sys.path.insert(0, str(package))
from ear_clipping import triangulate, triangle_area, polygon_area
```

## 1. Choose a concave boundary

The inward notch makes this different from a convex polygon. Change the coordinates, but keep the ring simple.

```python
ring = [(0, 0), (6, 0), (6, 5), (4, 2), (2, 4), (0, 5)]
triangles = triangulate(ring)
print(f'{len(ring)} input vertices → {len(triangles)} triangles')
print(f'Polygon area: {polygon_area(ring)}; triangle area sum: {sum((triangle_area(t) for t in triangles), start=0)}')
```

## 2. Inspect the triangulation

Each triangle is drawn separately so you can see the internal diagonals.

```python
fig, ax = plt.subplots(figsize=(7, 6))
for i, tri in enumerate(triangles):
    closed = list(tri) + [tri[0]]
    xs, ys = zip(*closed)
    ax.fill(xs, ys, alpha=0.28, edgecolor='black', linewidth=1.5)
    ax.text(sum(p[0] for p in tri)/3, sum(p[1] for p in tri)/3, str(i+1), ha='center', va='center')
bx, by = zip(*(ring + [ring[0]]))
ax.plot(bx, by, linewidth=2)
ax.set_aspect('equal'); ax.grid(True); ax.set_title('Ear-clipping result')
plt.show()
```

## 3. Check the invariants

For a simple polygon with n normalized vertices, a triangulation contains n − 2 triangles. Area agreement is a useful diagnostic, though topology and the ear tests are still needed to establish a valid partition.

```python
assert len(triangles) == len(ring) - 2
assert sum((triangle_area(t) for t in triangles), start=0) == polygon_area(ring)
print('Triangle count and area checks passed.')
```
