# Hidden Markov models and Viterbi decoding lab
Use a small discrete hidden Markov model to decode the most likely hidden-state sequence from observed symbols.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'implementations/python/hmm-viterbi'))
from hmm_viterbi import viterbi_decode
states = ('Rainy', 'Sunny')
symbols = ('walk', 'shop', 'clean')
start = (0.6, 0.4)
transition = ((0.7, 0.3), (0.4, 0.6))
emission = ((0.1, 0.4, 0.5), (0.6, 0.3, 0.1))
observations = ('walk', 'shop', 'clean')
path, log_probability = viterbi_decode(observations, states, symbols, start, transition, emission)
print('observations:', observations)
print('most likely hidden states:', path)
print('joint log probability:', log_probability)
```
