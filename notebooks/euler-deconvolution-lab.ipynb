# Euler deconvolution laboratory
Recover a synthetic source location from a field and its derivatives using the homogeneity equation. This controlled example validates the algebra; it is not a geological inversion workflow.

```python
import sys
sys.path.insert(0, '../implementations/python/potential-field')
import numpy as np
from spectral_derivatives import euler_deconvolution_window
rng=np.random.default_rng(42)
x,y=np.meshgrid(np.linspace(-2,2,8),np.linspace(-2,2,7))
tx,ty,tz=(rng.normal(size=x.shape) for _ in range(3))
truth={'x0':1.25,'y0':-0.75,'z0':3.5,'base_level':12.0}
structural_index=2.0; observation_z=0.0
field=((truth['x0']-x)*tx+(truth['y0']-y)*ty+(truth['z0']-observation_z)*tz)/structural_index+truth['base_level']
estimate=euler_deconvolution_window(field,tx,ty,tz,x,y,observation_z,structural_index)
estimate,truth
```

## Inspect the result
The synthetic derivatives are deliberately generated so the Euler equation holds exactly. Real derivative grids do not enjoy this guarantee. Explore smaller windows, noise, incorrect structural indices and multiple-source contamination. A low residual does not prove that a geological source model is correct. Rank-deficient windows are rejected.
