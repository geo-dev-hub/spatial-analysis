# Space and time can interact

Compare a separable covariance with one Gneiting-class nonseparable model. Data are synthetic. The implementation is restricted to Euclidean space and linear time.


```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
root=next((p for p in (Path.cwd().resolve(),*Path.cwd().resolve().parents) if (p/'implementations/python/nonseparable-space-time-covariance').is_dir()),None)
if root is None: raise RuntimeError('Open from a repository checkout')
sys.path.insert(0,str(root/'implementations/python/nonseparable-space-time-covariance'))
from covariance import gneiting_covariance,gneiting_covariance_matrix,covariance_matrix_diagnostics
```

## 1. Compare spatial profiles
Temporal lag changes the amplitude and effective spatial scale.


```python
h=np.linspace(0,5,150)
fig,ax=plt.subplots(figsize=(8,4.5))
for u in [0.,1.,3.]: ax.plot(h,gneiting_covariance(h,u,spatial_scale=1.5,temporal_scale=1,alpha=.8,beta=1),label=f'time lag {u:g}')
ax.set(xlabel='Spatial separation',ylabel='Covariance',title='Spatial profile changes with temporal lag'); ax.legend(); ax.grid(True,alpha=.25); plt.show()
```

## 2. Change the interaction parameter
At beta=0 this parameterisation has no temporal decay.


```python
for beta in [0.,.4,1.]: print(beta,gneiting_covariance(np.array([0.,1.,2.]),2.,beta=beta))
```

## 3. Build and inspect a covariance matrix
Each row is an observation with a location and a time.


```python
xy=np.array([[0,0],[1,0],[0,1],[2,1],[3,0],[2,3]],float); times=np.array([0,.5,1,2,2.5,4])
K=gneiting_covariance_matrix(xy,times,spatial_scale=2,temporal_scale=1.5,alpha=.8,beta=.7)
print(np.round(K,3)); print(covariance_matrix_diagnostics(K))
fig,ax=plt.subplots(figsize=(6,5)); im=ax.imshow(K); ax.set(title='Space-time covariance matrix',xlabel='Observation',ylabel='Observation'); fig.colorbar(im,ax=ax); plt.show()
```
