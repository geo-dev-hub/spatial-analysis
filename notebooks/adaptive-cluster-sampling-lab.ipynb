# Adaptive cluster sampling laboratory
Explore how a threshold-positive unit recruits its neighbours, and how recruitment can propagate through a network. The sample expansion is not itself a population estimator.

```python
import sys
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'implementations/python/adaptive-sampling'))
from adaptive_sampling import adaptive_cluster_sample
values=[0,4,0,6,0,5,0,0]
graph=[[1],[0,2],[1,3],[2,4],[3,5],[4,6],[5,7],[6]]
result=adaptive_cluster_sample(values,graph,[0,7],threshold=5)
result
```

```python
import matplotlib.pyplot as plt
positions=[(i,0.15*(i%2)) for i in range(len(values))]
plt.figure(figsize=(9,2.5))
for i,row in enumerate(graph):
    for j in row:
        if j>i:plt.plot([positions[i][0],positions[j][0]],[positions[i][1],positions[j][1]],'k-',alpha=.4)
for i,(x,y) in enumerate(positions):
    colour='tab:red' if i in result.initial_indices else ('tab:orange' if i in result.recruited_indices else 'tab:blue')
    plt.scatter(x,y,s=180,c=colour);plt.text(x,y+.08,str(i),ha='center')
plt.yticks([]);plt.title('Initial sample (red), recruited units (orange), unsampled (blue)');plt.grid(axis='x')
```

## Interpretation
The threshold rule triggers network expansion recursively. A recruited unit that does not meet the threshold is still part of the sample, but does not recruit further neighbours. Estimation requires the initial sampling design and correct inclusion probabilities; this notebook only demonstrates the expansion algorithm.
