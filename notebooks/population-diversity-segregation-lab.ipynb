# Population diversity and segregation laboratory
Compare overall diversity with multigroup segregation and group exposure using two contrasting synthetic population matrices.

```python
import sys, numpy as np
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'implementations/python/population-diversity'))
from population_diversity import shannon_entropy,simpson_concentration,reardon_firebaugh_entropy_index,exposure_matrix
separated=np.array([[90,10],[10,90],[80,20],[20,80]])
mixed=np.array([[50,50],[50,50],[50,50],[50,50]])
```

```python
for name,counts in [('Separated',separated),('Mixed',mixed)]:
    print(name,'segregation H=',reardon_firebaugh_entropy_index(counts))
    print('overall Shannon=',shannon_entropy(counts.sum(axis=0)))
    print('unit Shannon=',shannon_entropy(counts,axis=1))
    print('exposure=',exposure_matrix(counts))
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,2,figsize=(8,3))
for a,z,title in zip(ax,[separated,mixed],['Separated composition','Identical composition']):
    im=a.imshow(z,aspect='auto');a.set_title(title);a.set_xlabel('Group');a.set_ylabel('Area unit');fig.colorbar(im,ax=a)
plt.tight_layout()
```

## Interpretation
A high overall diversity value does not imply low segregation: the total population can be diverse while groups occupy different units. The exposure matrix is an areal-unit measure, not a distance-weighted spatial interaction model. Always inspect the unit system, group definitions and population denominators.
