Use the statsmodels reference implementation through the package wrapper.

```python
import sys
sys.path.insert(0, '../implementations/python/stl-decomposition')
from stl_decomposition import stl_decompose
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng=np.random.default_rng(7)
t=np.arange(180)
y=0.025*t + 1.8*np.sin(2*np.pi*t/12) + rng.normal(0,0.25,len(t))
y[91]+=3.5
parts=stl_decompose(y,period=12,seasonal=7,robust=True)
```

```python
fig,axes=plt.subplots(4,1,figsize=(11,8),sharex=True)
for ax,series,label in zip(axes,[y,parts['trend'],parts['seasonal'],parts['resid']],['Observed','Trend','Seasonal','Remainder']):
    ax.plot(t,series); ax.set_ylabel(label); ax.grid(True)
axes[-1].set_xlabel('Time step')
fig.tight_layout()
```

```python
reconstructed=np.array(parts['trend'])+parts['seasonal']+parts['resid']
float(np.max(np.abs(reconstructed-y)))
```
