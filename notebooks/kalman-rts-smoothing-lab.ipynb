# Kalman filter and RTS smoother laboratory
A synthetic 1D random-walk example. The forward filter uses observations up to the current time; the backward RTS smoother revises earlier states using future observations.

```python
import sys, numpy as np
sys.path.insert(0, '../implementations/python/curves-smoothing')
from kalman_smoothing import kalman_rts_smoother
rng=np.random.default_rng(7)
truth=np.cumsum(rng.normal(0,0.25,50))
observed=truth+rng.normal(0,0.8,50)
observed[18:22]=np.nan
```

```python
result=kalman_rts_smoother(observed,[[1.0]],[[1.0]],[[0.08]],[[0.64]],[0.0],[[1.0]])
filtered=result['filtered_means'][:,0]
smoothed=result['smoothed_means'][:,0]
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,4));plt.plot(truth,label='latent synthetic state');plt.scatter(range(len(observed)),observed,label='observations',s=14);plt.plot(filtered,label='filtered');plt.plot(smoothed,label='RTS smoothed');plt.legend();plt.grid(True)
```

## Interpretation
The filter is causal: it does not use future measurements. The RTS smoother is retrospective: it uses the full observation sequence to revise earlier estimates. The gap demonstrates prediction-only filtering for missing observations. Change Q and R to explore the balance between process variation and measurement noise. This example does not establish a suitable motion model or noise level for real GNSS trajectories.
