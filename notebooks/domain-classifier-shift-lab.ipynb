# Domain-classifier distribution-shift lab
Train a classifier to distinguish source from target predictor samples. Out-of-fold ROC AUC measures how distinguishable the two domains are under this model and feature representation.

```python
import sys
sys.path.insert(0, '../implementations/python/domain-classifier-shift')
import numpy as np
from domain_classifier import domain_classifier_diagnostic
rng = np.random.default_rng(19)
source = rng.normal(loc=0, scale=1, size=(400, 3))
target = rng.normal(loc=[.5, 0, 1.0], scale=1, size=(400, 3))
result = domain_classifier_diagnostic(source, target, n_splits=5)
print('Out-of-fold ROC AUC:', result['roc_auc'])
print('Balanced accuracy:', result['balanced_accuracy'])
```

```python
groups = np.repeat(np.arange(40), 20)
grouped = domain_classifier_diagnostic(source, target, n_splits=5, groups=groups)
print('Grouped out-of-fold ROC AUC:', grouped['roc_auc'])
print('Splitter:', grouped['splitter'])
```

## Interpretation limits
AUC near 0.5 means this logistic-regression classifier has difficulty distinguishing the domains; it does not prove equality. A high AUC does not identify the cause of shift or guarantee model error. Use spatial/temporal groups that reflect independent deployment units. This is not a calibrated two-sample p-value and cannot detect concept shift from predictors alone.
