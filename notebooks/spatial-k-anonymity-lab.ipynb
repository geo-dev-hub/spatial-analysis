# Spatial k-anonymity lab
This lab generalises a synthetic set of point locations into disjoint quadtree cells, each containing at least k records, or suppresses records when the full population is too small.

```python
import sys
sys.path.insert(0, '../implementations/python/spatial-privacy')
import numpy as np
import matplotlib.pyplot as plt
from spatial_k_anonymity import spatial_k_anonymity
rng = np.random.default_rng(8)
points_a = rng.normal(loc=(0, 0), scale=.08, size=(35, 2))
points_b = rng.normal(loc=(1, 1), scale=.10, size=(45, 2))
points = np.vstack([points_a, points_b])
result = spatial_k_anonymity(points[:, 0], points[:, 1], k=5, max_depth=7)
print('Groups:', len(result['groups']), 'suppressed:', int(result['suppressed'].sum()))
print('Group counts:', [g['count'] for g in result['groups']])
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(points[:, 0], points[:, 1], s=12, alpha=.35, label='Input points')
for group in result['groups']:
    x0, y0, x1, y1 = group['bounds']
    rect = plt.Rectangle((x0, y0), x1-x0, y1-y0, fill=False, linewidth=1.5)
    ax.add_patch(rect)
ax.set_aspect('equal'); ax.set_title('Generalised quadtree cells (each count >= k)'); ax.legend(); plt.show()
```

## Interpretation limits
This protects only the location equivalence classes formed from the supplied records. Other released attributes, temporal traces, auxiliary data, population definition and overlapping releases can still enable re-identification. It is not differential privacy. Use a projected coordinate system and review the complete release under a threat model.
