Calculate empirical ensemble CRPS and an expected rank histogram. Ties are fractionally distributed to avoid arbitrary ordering.

```python
import sys
sys.path.insert(0, '../implementations/python/forecast-verification')
from ensemble_verification import crps_ensemble, mean_crps_ensemble, expected_rank_histogram
observed=[0.2,1.1,2.0,1.4,0.0]
ensemble=[[0.0,0.3,0.4],[0.5,1.0,1.4],[1.2,1.8,2.4],[0.8,1.3,1.9],[-0.2,0.1,0.5]]
scores=crps_ensemble(observed,ensemble)
scores, mean_crps_ensemble(observed,ensemble)
```

```python
hist=expected_rank_histogram(observed,ensemble)
hist
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4));plt.bar(range(len(hist)),hist);plt.xlabel('Ensemble rank of observation');plt.ylabel('Expected count');plt.title('Expected rank histogram');plt.grid(axis='y')
```

## Interpretation
CRPS rewards both proximity to the observation and useful ensemble spread. Rank histograms diagnose the distribution of observations relative to ensemble members. Neither should be interpreted without sample size, dependence, stratification and a proper verification design.
