# Discrete p-center

Choose exactly p sites to minimize the worst nearest-facility cost. This minimax objective differs from p-median's total cost and MCLP's threshold coverage.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'implementations/python/p-center'))
import numpy as np
import matplotlib.pyplot as plt
from p_center import solve_p_center
```

```python
costs = np.array([[4, 18, 22, 12, 16], [7, 14, 19, 9, 13], [16, 5, 11, 17, 12], [20, 8, 9, 15, 7], [12, 16, 4, 10, 18], [17, 11, 8, 13, 5]])
for p in (1, 2, 3):
    sites, radius, assignment = solve_p_center(costs, p)
    print(f'p={p}: sites={sites}, worst cost={radius}, assignments={assignment}')
```

## Inspect one optimal solution

The maximum of the nearest-site costs is the p-center objective. The highlighted demand point is the one setting the radius.

```python
p = 2
sites, radius, assignment = solve_p_center(costs, p)
nearest = costs[np.arange(costs.shape[0]), assignment]
worst = int(np.argmax(nearest))
fig, ax = plt.subplots(figsize=(8,4))
im = ax.imshow(costs, aspect='auto')
ax.set_xticks(range(costs.shape[1]), [f'Site {j}' for j in range(costs.shape[1])])
ax.set_yticks(range(costs.shape[0]), [f'Demand {i}' for i in range(costs.shape[0])])
ax.set_title(f'p-center: p={p}, worst assignment cost={radius}')
ax.set_xlabel('Candidate facilities'); ax.set_ylabel('Demand locations')
for i in range(costs.shape[0]):
    for j in range(costs.shape[1]): ax.text(j,i,f'{costs[i,j]:.0f}',ha='center',va='center')
ax.text(0.01, -0.22, f'Worst-served demand: {worst} → site {assignment[worst]} (cost {nearest[worst]})', transform=ax.transAxes)
fig.colorbar(im, ax=ax, label='Cost units'); plt.tight_layout(); plt.show()
```

## Interpretation and limits

A worst-case objective can be dominated by one remote or atypical demand point. Check whether that point belongs in the planning population and whether the cost matrix reflects the actual service mode. The baseline has no capacity, queueing, uncertainty or equity constraints.
