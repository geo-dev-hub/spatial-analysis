# AIS reporting-gap detection lab
Create two synthetic vessel tracks, sort reports by vessel/time, and flag long intervals or high implied speeds. The result is a screening signal, not proof of deliberate AIS interruption.

```python
import sys
sys.path.insert(0, '../implementations/python/ais-gap-detection')
import numpy as np
from ais_gap_detection import detect_ais_gaps
vessel = ['A','A','A','B','B','B']
time_s = [0, 600, 4200, 0, 600, 1200]
lon = [0, .08, .2, 10, 10.08, 10.16]
lat = [0, 0, 0, 0, 0, 0]
findings = detect_ais_gaps(vessel, time_s, lon, lat, max_gap_seconds=1800, max_speed_knots=40)
for finding in findings: print(finding)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(8, 3))
for name in sorted(set(vessel)):
    idx = [i for i, value in enumerate(vessel) if value == name]
    order = sorted(idx, key=lambda i: time_s[i])
    ax.plot(np.array(lon)[order], np.array(lat)[order], marker='o', label=f'Vessel {name}')
for item in findings:
    ax.plot([lon[item['from_index']], lon[item['to_index']]], [lat[item['from_index']], lat[item['to_index']]], linestyle='--', linewidth=3)
ax.set_xlabel('Longitude (degrees)'); ax.set_ylabel('Latitude (degrees)'); ax.set_title('Dashed segments are flagged intervals'); ax.legend(); plt.show()
```

## Review before operational use
Thresholds must reflect vessel class, AIS reporting mode, region and reception coverage. A missing report does not establish intentional concealment. High implied speed can result from bad timestamps, position errors or identity collisions.
