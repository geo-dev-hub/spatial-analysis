# Spatial uncertainty sampling laboratory
Select a batch of candidate map-validation sites using uncertainty per unit cost, while enforcing a minimum distance between selected locations.

```python
import sys, numpy as np
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'implementations/python/adaptive-sampling'))
from active_learning import select_uncertain_spatial_sites
rng=np.random.default_rng(4)
xy=rng.uniform(0,10,size=(80,2))
uncertainty=np.exp(-((xy[:,0]-6)**2+(xy[:,1]-5)**2)/12)+.15*rng.random(80)
cost=1+.1*xy[:,0]
result=select_uncertain_spatial_sites(xy,uncertainty,8,min_distance=1.5,costs=cost)
selected=result['selected_indices']
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,6));plt.scatter(xy[:,0],xy[:,1],c=uncertainty/cost,s=35,label='candidate utility');plt.colorbar(label='uncertainty / cost')
plt.scatter(xy[selected,0],xy[selected,1],marker='*',s=180,edgecolors='black',label='selected batch');plt.axis('equal');plt.legend();plt.title('Greedy spatially diverse uncertainty sampling');plt.grid(True)
```

## Interpretation
The selector ranks candidates by uncertainty per cost and greedily enforces a minimum separation. This avoids a batch concentrated in one hotspot, but it is not a globally optimal design. A full active-learning loop would obtain labels, update the model, recompute uncertainty and repeat. Inclusion probabilities and unbiased map-accuracy estimates require a separate sampling design.
