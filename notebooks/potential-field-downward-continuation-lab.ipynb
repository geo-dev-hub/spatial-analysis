# Regularized downward continuation lab
Downward continuation is an ill-posed inverse: the unregularized Fourier multiplier grows exponentially with wavenumber. This lab demonstrates a Tikhonov-stabilized inverse and its resolution/noise trade-off.

```python
import sys
sys.path.insert(0, '../implementations/python/potential-field')
import numpy as np
import matplotlib.pyplot as plt
from spectral_derivatives import upward_continuation, downward_continuation_regularized
n = 128
x = np.arange(n)
k_low, k_high = 2*np.pi/n, np.pi/4
surface = np.tile(2 + np.cos(k_low*x) + 0.2*np.cos(k_high*x), (n, 1))
height = 3.0
observed = upward_continuation(surface, 1, 1, height)
rng = np.random.default_rng(12)
noisy = observed + rng.normal(0, 0.003, observed.shape)
recovered = downward_continuation_regularized(noisy, 1, 1, height, regularization=1e-5)
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, field, title in zip(axes, [surface, noisy, recovered], ['Original surface', 'Upward continued + noise', 'Regularized downward result']):
    im = ax.imshow(field, origin='lower'); ax.set_title(title); fig.colorbar(im, ax=ax)
plt.show()
print('RMSE before:', np.sqrt(np.mean((noisy-surface)**2)))
print('RMSE after:', np.sqrt(np.mean((recovered-surface)**2)))
```

## Interpretation limits
The regularisation parameter is a modelling choice. Too little regularisation amplifies high-frequency noise; too much suppresses genuine short-wavelength signal. The zero-frequency mean is preserved, but other modes are filtered. This is not an automatic or unique recovery. FFT periodic-boundary assumptions and grid projection requirements still apply.
