# UCB1 survey-arm allocation laboratory
Simulate a stationary multi-armed bandit. Each arm represents a candidate survey action with an independent reward distribution. This is a generic baseline, not a spatially correlated bandit.

```python
import sys
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'implementations/python/adaptive-sampling'))
from ucb1 import run_ucb1
result=run_ucb1([0.1,0.4,0.9,0.2],budget=250,noise_std=.25,random_state=11)
result['counts'],result['estimated_means']
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4));plt.bar(range(len(result['counts'])),result['counts']);plt.xlabel('Arm');plt.ylabel('Number of pulls');plt.title('UCB1 allocation');plt.grid(axis='y')
```

## Interpretation
UCB1 initially samples every arm, then balances empirical reward against an exploration bonus. Its assumptions are independent stationary arms. Candidate locations with correlated rewards violate that abstraction; a spatial/contextual bandit needs a model that shares information across locations.
