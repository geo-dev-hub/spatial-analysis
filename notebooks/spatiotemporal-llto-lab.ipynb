# Leave-location-and-time-out validation lab
A synthetic panel demonstrates train, test and buffer rows in a strict LLTO fold.

```python
import sys
sys.path.insert(0, '../implementations/python/spatiotemporal-validation')
import numpy as np
import matplotlib.pyplot as plt
from spatiotemporal_validation import leave_location_and_time_out
locations = np.repeat(['North', 'Central', 'South'], 4)
times = np.tile([1, 2, 3, 4], 3)
folds = leave_location_and_time_out(locations, times)
fold = next(f for f in folds if f['location'] == 'Central' and f['time'] == 2)
print('Train:', fold['train'], 'Test:', fold['test'], 'Buffer:', fold['buffer'])
```

```python
state = np.zeros((3, 4), dtype=int)
for i in fold['buffer']:
    state[{'North':0,'Central':1,'South':2}[locations[i]], times[i]-1] = 1
for i in fold['test']:
    state[{'North':0,'Central':1,'South':2}[locations[i]], times[i]-1] = 2
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(state, aspect='auto')
ax.set_yticks(range(3), ['North', 'Central', 'South'])
ax.set_xticks(range(4), [1, 2, 3, 4])
ax.set_xlabel('Time group'); ax.set_title('One LLTO fold: train=0, buffer=1, test=2')
plt.show()
```

## Interpretation
Only the Central-at-time-2 cell is tested. The entire Central location and all locations at time 2 are withheld from training; the cross-shaped remainder is the buffer. This evaluates joint transfer, not future-only forecasting or distance-based spatial extrapolation.
