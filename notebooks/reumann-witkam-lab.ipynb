# Reumann–Witkam line simplification lab
Establish a strip, scan vertices, retain the vertex immediately before a strip exit, and repeat. Coordinates are planar; tolerance uses the same units.

```python
from pathlib import Path
import sys
root = Path.cwd()
for candidate in (root, *root.parents):
    package = candidate / 'implementations' / 'python' / 'reumann-witkam'
    if package.exists():
        sys.path.insert(0, str(package))
        break
from simplify import reumann_witkam
```

```python
line = [(0, 0), (1, 0.05), (2, -0.03), (3, 0), (4, 0.8), (5, 0.82), (6, 0.1)]
simplified = reumann_witkam(line, tolerance=0.2)
print('Input vertices:', len(line))
print('Output vertices:', len(simplified))
print('Simplified:', simplified)
```

```python
for tolerance in (0.05, 0.2, 0.5, 1.0):
    result = reumann_witkam(line, tolerance)
    print(f'tolerance={tolerance:>4}: {len(result)} vertices -> {result}')
```
