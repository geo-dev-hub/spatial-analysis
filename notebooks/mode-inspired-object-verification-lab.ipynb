# MODE-inspired object verification lab
Extract coherent regions from synthetic forecast and observed fields, measure attributes and perform transparent one-to-one matching. This is not a full MET/MODE implementation.

```python
import sys
sys.path.insert(0, '../implementations/python/forecast-verification')
import numpy as np
import matplotlib.pyplot as plt
from object_verification import identify_objects, match_objects
yy, xx = np.mgrid[:60, :70]
forecast = 8*np.exp(-((xx-24)**2+(yy-25)**2)/(2*5**2)) + 6*np.exp(-((xx-52)**2+(yy-42)**2)/(2*3**2))
observed = 8*np.exp(-((xx-27)**2+(yy-25)**2)/(2*5**2))
f = identify_objects(forecast, threshold=1.5, smooth_sigma=1.0, min_area=5)
o = identify_objects(observed, threshold=1.5, smooth_sigma=1.0, min_area=5)
pairs = match_objects(f['objects'], o['objects'], forecast.shape, max_centroid_distance=0.2)
print('Forecast objects:', f['objects'])
print('Observed objects:', o['objects'])
print('Matches:', pairs['matches'])
print('Unmatched forecast:', pairs['unmatched_forecast'])
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(f['labels']); ax[0].set_title('Forecast objects')
ax[1].imshow(o['labels']); ax[1].set_title('Observed objects')
plt.show()
```

## Interpretation
Object counts and matches depend on smoothing, threshold, minimum area, connectivity and the matching cost. The matching rule here is intentionally simpler than MODE's configurable fuzzy-interest and merge/cluster workflow.
