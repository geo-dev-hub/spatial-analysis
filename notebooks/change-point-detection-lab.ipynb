# Change-point detection laboratory
Compare online CUSUM with greedy binary segmentation and penalised PELT for piecewise-constant mean shifts. This is a controlled synthetic experiment, not an environmental validation dataset.

```python
import sys
sys.path.insert(0, '../implementations/python/change-point-detection')
from change_points import cusum, binary_segmentation, pelt_mean
series = [0.0]*30 + [2.5]*25 + [-1.0]*25
series[:5], series[-5:]
```

```python
online_alarms = cusum(series, target_mean=0.0, allowance=0.25, threshold=3.0)
greedy = binary_segmentation(series, penalty=3.0, min_size=5)
optimal = pelt_mean(series, penalty=5.0, min_size=5)
online_alarms, greedy, optimal
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,4)); plt.plot(series, marker='.', linewidth=1)
for cp in greedy: plt.axvline(cp, linestyle='--', alpha=.6)
for cp in optimal: plt.axvline(cp, linestyle=':', alpha=.8)
plt.xlabel('Observation index'); plt.ylabel('Value'); plt.title('Greedy (dashed) and PELT (dotted) change points'); plt.grid(True)
```

## Interpretation
CUSUM is sequential and depends on a known reference mean and calibrated thresholds. Binary segmentation makes greedy recursive splits. PELT optimises an additive penalised SSE objective for this implementation's stated assumptions. A detected shift does not identify its cause. Add seasonality, serial dependence, missingness and spatial multiplicity before applying these methods to Earth-observation data.
