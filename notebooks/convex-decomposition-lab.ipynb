# Convex decomposition: triangulate, then merge

This lab shows how a simple polygon can be partitioned into convex pieces. The implementation is a transparent greedy teaching baseline; it does not claim an optimal partition or the Hertel–Mehlhorn approximation guarantee.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT != ROOT.parent and not (ROOT/'implementations').exists(): ROOT=ROOT.parent
sys.path.insert(0,str(ROOT/'implementations/python/convex-decomposition'))
sys.path.insert(0,str(ROOT/'implementations/python/ear-clipping'))
from convex_decomposition import convex_decompose
import matplotlib.pyplot as plt
```

## 1. Start with a concave polygon

The L-shaped polygon below has a reflex corner. Triangulation is always a valid convex partition, but merging adjacent triangles can reduce the number of pieces.

```python
polygon=[(0,0),(3,0),(3,1),(1,1),(1,3),(0,3)]
pieces=convex_decompose(polygon)
print(f'{len(pieces)} convex pieces')
pieces
```

```python
fig,ax=plt.subplots(figsize=(7,5))
for i,piece in enumerate(pieces):
    ring=piece+[piece[0]]
    xs=[float(p[0]) for p in ring]; ys=[float(p[1]) for p in ring]
    ax.fill(xs,ys,alpha=.25,label=f'piece {i+1}')
    ax.plot(xs,ys,marker='o')
ax.set_aspect('equal',adjustable='box'); ax.grid(True); ax.legend()
ax.set_title('Greedy convex partition of an L-shaped polygon')
plt.show()
```

## 2. Compare with a convex input

A convex polygon can be triangulated and then merged back into one piece. This is a useful sanity check, not evidence that greedy merging finds a globally minimum partition for every concave polygon.

```python
rectangle=[(0,0),(4,0),(4,3),(0,3)]
rectangle_pieces=convex_decompose(rectangle)
len(rectangle_pieces), rectangle_pieces
```
