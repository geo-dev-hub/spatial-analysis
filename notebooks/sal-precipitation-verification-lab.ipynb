# SAL precipitation verification lab
Compute Structure, Amplitude and Location components on synthetic forecast/observation fields. Objects are independently thresholded; the SAL method does not require forecast-observation object pairing.

```python
import sys
sys.path.insert(0, '../implementations/python/forecast-verification')
import numpy as np
import matplotlib.pyplot as plt
from sal_verification import sal_score
yy, xx = np.mgrid[:50, :60]
obs = 8*np.exp(-((xx-20)**2+(yy-24)**2)/(2*4**2)) + 4*np.exp(-((xx-43)**2+(yy-12)**2)/(2*3**2))
fcst = 1.4*(8*np.exp(-((xx-23)**2+(yy-24)**2)/(2*5**2)) + 4*np.exp(-((xx-46)**2+(yy-12)**2)/(2*4**2)))
scores = sal_score(fcst, obs, threshold_fraction=1/15, connectivity=8)
print({k: scores[k] for k in ['structure','amplitude','location','location_l1','location_l2']})
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(fcst); ax[0].set_title('Forecast')
ax[1].imshow(obs); ax[1].set_title('Observation')
plt.show()
```

## Interpretation
Positive A indicates forecast over-amplitude; negative A indicates under-amplitude. Positive S indicates larger/flatter objects in forecast under this convention. L combines normalized displacement of total centres of mass and difference in object scatter. The object extraction threshold and connectivity are part of the metric contract.
