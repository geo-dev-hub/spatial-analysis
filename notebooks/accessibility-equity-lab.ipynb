# Accessibility distribution and equity lab

Synthetic accessibility values only. This lab does not calculate routes or decide what is fair.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'implementations' / 'python' / 'accessibility-equity'))
import matplotlib.pyplot as plt
from equity_diagnostics import accessibility_summary, group_accessibility_summary, group_mean_gap, weighted_gini, weighted_lorenz_curve
```

```python
names = ['A', 'B', 'C', 'D']
access = [2, 4, 8, 10]
population = [100, 300, 200, 100]
groups = ['outer', 'outer', 'inner', 'inner']
minimum_standard = 5
print('Place-weighted:', accessibility_summary(access, minimum_threshold=minimum_standard))
print('Population-weighted:', accessibility_summary(access, population, minimum_standard))
print('Gini:', weighted_gini(access), weighted_gini(access, population))
```

```python
p0, l0 = weighted_lorenz_curve(access)
p1, l1 = weighted_lorenz_curve(access, population)
plt.figure(figsize=(7, 6))
plt.plot([0, 1], [0, 1], linestyle='--', label='Equal distribution')
plt.plot(p0, l0, marker='o', label='Each place equally weighted')
plt.plot(p1, l1, marker='o', label='Population weighted')
plt.xlabel('Cumulative population / place share')
plt.ylabel('Cumulative accessibility share')
plt.title('Lorenz curves under two weighting choices')
plt.legend(); plt.grid(alpha=0.25); plt.show()
```

```python
by_group = group_accessibility_summary(access, groups, population)
print(by_group)
print(group_mean_gap(by_group))
```
