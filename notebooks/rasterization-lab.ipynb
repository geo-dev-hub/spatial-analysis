# Rasterisation lab
Compare thin-line rasterisers, conservative traversal, anti-aliasing and polygon scan conversion. The package uses grid coordinates, not geographic coordinates.


## 1. Load the teaching implementation
Run from the repository root. The notebook adds the package folder to the import path.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'implementations/python/rasterization'))
from rasterization import dda_line, bresenham_line, supercover_line, wu_line, scanline_polygon_fill
```

## 2. Thin line versus conservative coverage


```python
start, end = (0, 0), (6, 3)
print('DDA:', dda_line(start, end))
print('Bresenham:', bresenham_line(start, end))
print('Supercover:', supercover_line(start, end))
```

A supercover intentionally returns more cells than a thin-line rasteriser. It is useful when missing a touched cell would be unsafe, such as conservative grid traversal.
## 3. Anti-aliased intensity


```python
wu_line((0, 0), (6, 3))
```

Wu weights are display intensities. They are not fractional polygon area coverage and should not be interpreted as probabilities.
## 4. Polygon scan conversion


```python
ring = [(0, 0), (5, 0), (5, 4), (0, 4)]
filled = scanline_polygon_fill(ring)
print('filled cell count:', len(filled))
print('cells:', sorted(filled, key=lambda cell: (cell[1], cell[0])))
```
