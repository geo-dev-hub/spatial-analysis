Explore a user-level Laplace mechanism for counts over a fixed cell domain. The contribution bound determines sensitivity; repeated releases compose privacy loss.

```python
import sys
sys.path.insert(0, '../implementations/python/spatial-privacy')
from differential_privacy import private_spatial_counts
records=[('person-a','cell-1'),('person-a','cell-2'),('person-b','cell-2'),('person-c','cell-3'),('person-c','cell-3')]
release=private_spatial_counts(records,['cell-1','cell-2','cell-3','cell-4'],epsilon=1.0,max_cells_per_person=1,rng=17)
release
```

## Inspect the guarantee
With at most k cells contributed per person, add/remove-one-person L1 sensitivity is k and Laplace noise scale is k/epsilon. Try k=2 and observe the larger noise scale. This is one release only: multiple releases require composition accounting. The mechanism does not protect against incorrectly linking one person's records or choosing an unbounded contribution policy.
