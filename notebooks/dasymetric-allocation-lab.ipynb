# Weighted dasymetric allocation lab
Allocate two synthetic source-zone totals to smaller target units using building-area proxy weights.

```python
import sys
sys.path.insert(0, '../implementations/python/dasymetric-allocation')
import numpy as np
import matplotlib.pyplot as plt
from dasymetric_allocation import dasymetric_allocate
source_population = np.array([1200, 800])
building_area = np.array([[20, 60, 20, 0], [0, 10, 30, 60]], dtype=float)
result = dasymetric_allocate(source_population, building_area)
print('Allocated target totals:', result['target_totals'])
print('Conserved source totals:', result['source_totals_reconstructed'])
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(np.arange(4), result['target_totals'])
ax.set_xlabel('Target cell'); ax.set_ylabel('Allocated population')
ax.set_title('Population allocated by ancillary building-area weights')
plt.show()
```

## Interpretation
The totals are conserved by construction. The target-level pattern is only as credible as the ancillary weights and source-zone geometry. This lab does not perform polygon overlays or validate population estimates.
