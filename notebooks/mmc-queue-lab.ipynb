# Queueing at a service facility: M/M/c

Explore how arrival demand and parallel service capacity affect waiting. This is a steady-state model with Poisson arrivals, exponential service times, identical servers and one FCFS queue. It is not a time-varying simulation.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / 'implementations/python/mmc-queue'))
import pandas as pd
import matplotlib.pyplot as plt
from mmc_queue import mmc_metrics
```

```python
arrival_rate = 8.0  # customers per hour
service_rate = 5.0  # customers per hour, per server
servers = 2
metrics = mmc_metrics(arrival_rate, service_rate, servers)
pd.Series(metrics, name='M/M/c result')
```

## Compare staffing levels

Keep arrival and per-server service rates fixed. Add servers and inspect how the probability of waiting and expected queue wait respond.

```python
rows = []
for c in (1, 2, 3, 4):
    if arrival_rate < c * service_rate:
        m = mmc_metrics(arrival_rate, service_rate, c)
        rows.append({'servers': c, 'utilisation': m['utilisation'],
                     'probability of waiting': m['probability_wait'],
                     'expected queue wait (hours)': m['expected_wait'],
                     'expected queue length': m['expected_queue_length']})
pd.DataFrame(rows)
```

## Approach the stability boundary

For a fixed server count, increase the arrival rate toward total service capacity. The model rejects an unstable queue rather than reporting a finite steady-state wait.

```python
arrival_rates = [5, 6, 7, 8, 9, 9.5]
waits = [mmc_metrics(rate, service_rate, servers)['expected_wait'] for rate in arrival_rates]
plt.figure(figsize=(7, 4))
plt.plot(arrival_rates, waits, marker='o')
plt.xlabel('Arrivals per hour')
plt.ylabel('Expected queue wait (hours)')
plt.title('M/M/c waiting rises near total service capacity')
plt.grid(True)
plt.show()
```
