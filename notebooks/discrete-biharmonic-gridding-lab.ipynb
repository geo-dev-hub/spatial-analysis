# Discrete biharmonic minimum-curvature lab
Fill missing interior nodes while holding the perimeter and any supplied interior observations fixed.

```python
import sys
sys.path.insert(0, '../implementations/python/geophysical-gridding')
import numpy as np
import matplotlib.pyplot as plt
from discrete_biharmonic import discrete_biharmonic_grid
x, y = np.meshgrid(np.linspace(-2, 2, 21), np.linspace(-2, 2, 21))
truth = np.exp(-(x*x + y*y)) + .15*x - .1*y
observed = truth.copy()
observed[1:-1, 1:-1] = np.nan
observed[5, 7] = truth[5, 7]
grid = discrete_biharmonic_grid(observed, dx=x[0,1]-x[0,0], dy=y[1,0]-y[0,0])
print('Maximum absolute reconstruction error:', np.max(np.abs(grid-truth)))
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(grid, origin='lower', extent=[x.min(), x.max(), y.min(), y.max()])
ax.set_title('Discrete biharmonic interpolation'); ax.set_xlabel('x'); ax.set_ylabel('y')
fig.colorbar(im, ax=ax, label='Interpolated value'); plt.show()
```
