# Stream order: comparing Strahler and Shreve

Stream order depends on the branching structure of a channel network. This lab starts with a synthetic, explicitly defined D8 stream mask. It does not extract streams from a DEM or choose a channel-initiation threshold.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'implementations').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'implementations/python/stream-order'))
from stream_order import stream_order
```

## 1. Construct a small branching network

Four headwater links form two second-order tributaries. Those tributaries meet downstream. Direction codes use the ESRI D8 convention.

```python
directions = np.array([
    [-1, -1, -1, 4, -1],
    [-1, 4, -1, 8, 16],
    [1, 1, 1, -1, -1],
    [-1, -1, 0, -1, -1],
] , dtype=int)
stream_mask = directions != -1
strahler, shreve = stream_order(directions, stream_mask)
print('Strahler order:')
print(strahler)
print('Shreve magnitude:')
print(shreve)
```

## 2. Compare the measures

Strahler order increases only when two or more tributaries of the same maximum order meet. Shreve magnitude adds the magnitudes of all incoming tributaries.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, values, title in [(axes[0], strahler, 'Strahler order'), (axes[1], shreve, 'Shreve magnitude')]:
    shown = np.ma.masked_where(~stream_mask, values)
    im = ax.imshow(shown, cmap='viridis')
    ax.set_title(title); ax.set_xticks(range(directions.shape[1])); ax.set_yticks(range(directions.shape[0]))
    for r, c in zip(*np.where(stream_mask)):
        ax.text(c, r, str(values[r, c]), ha='center', va='center', color='white')
    fig.colorbar(im, ax=ax)
plt.tight_layout(); plt.show()
```
