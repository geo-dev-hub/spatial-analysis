# Equity-constrained maximal covering lab

The model first enforces group-specific minimum coverage, then maximises total covered objective weight. Group labels and coverage floors are analyst-supplied policy choices.

```python
import sys
from pathlib import Path
relative_package = Path('implementations/python/equity-covering')
repo_root = next((candidate for candidate in (Path.cwd(), Path.cwd().parent)
                  if (candidate / relative_package).is_dir()), None)
if repo_root is None:
    raise FileNotFoundError(f'Cannot locate {relative_package} from {Path.cwd()}')
sys.path.insert(0, str(repo_root / relative_package))
from equity_covering import equity_maximal_covering
```

```python
costs = [[1, 9, 8], [1, 8, 9], [9, 1, 8], [9, 1, 8]]
groups = ['A', 'A', 'B', 'B']
ordinary = equity_maximal_covering(costs, groups, {'A': 0, 'B': 0}, 1, 2)
constrained = equity_maximal_covering(costs, groups, {'A': .5, 'B': .5}, 2, 2)
print('Ordinary:', ordinary)
print('Group floors:', constrained)
```

```python
floors = [0.0, 0.25, 0.5, 0.75, 1.0]
for floor in floors:
    result = equity_maximal_covering(costs, groups, {'A': floor, 'B': floor}, 2, 2)
    print(floor, result['selected_facilities'], result['objective'], result['group_coverage'])
```

## Separate population requirements from the objective

Change demand_weights to change each group's coverage denominator. Change objective_weights to alter which covered demand contributes to the objective. These are distinct modelling choices.

```python
weighted = equity_maximal_covering(
    costs, groups, {'A': .5, 'B': .5}, 2, 2,
    demand_weights=[100, 100, 50, 50],
    objective_weights=[1, 1, 5, 5],
)
print(weighted)
```
