# B-spline and NURBS laboratory
Evaluate polynomial and rational splines with de Boor's algorithm. The quarter-circle fixture demonstrates why rational weights matter.

```python
import sys, math
sys.path.insert(0, '../implementations/python/curves-smoothing')
from splines import bspline_curve, nurbs_curve, bspline_point, nurbs_point
controls=[(0,0),(1,2),(2,0)]; knots=[0,0,0,1,1,1]
polynomial=bspline_curve(controls,2,knots,51)
```

```python
circle_controls=[(1,0),(1,1),(0,1)]
weights=[1,math.sqrt(0.5),1]
circle=nurbs_curve(circle_controls,weights,2,knots,51)
nurbs_point(circle_controls,weights,2,knots,0.5)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,6))
plt.plot(*zip(*polynomial),label='quadratic B-spline / Bezier')
plt.plot(*zip(*circle),label='rational quadratic quarter-circle')
plt.scatter(*zip(*circle_controls),marker='x',label='circle control points')
plt.axis('equal');plt.grid(True);plt.legend()
```
