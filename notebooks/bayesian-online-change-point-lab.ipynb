# Bayesian online change-point detection lab
The run-length posterior tracks how long the current regime may have lasted. This implementation assumes an unknown segment mean, known Gaussian observation variance and constant hazard.

```python
import sys
sys.path.insert(0, '../implementations/python/change-point-detection')
from bayesian_online import bayesian_online_changepoint
series=[0.0]*25+[3.5]*20+[-1.0]*20
result=bayesian_online_changepoint(series,hazard=.03,observation_variance=.25,prior_mean=0,prior_variance=1)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,4));plt.plot(result['change_probability']);plt.xlabel('Observation index');plt.ylabel('Posterior P(change)');plt.title('Online change probability');plt.grid(True)
```

```python
import numpy as np
posterior=np.asarray(result['run_length_posterior'])
plt.figure(figsize=(10,5));plt.imshow(posterior.T,origin='lower',aspect='auto',interpolation='nearest');plt.xlabel('Time');plt.ylabel('Run length');plt.title('Posterior over run length');plt.colorbar(label='Probability')
```

## Interpretation
The posterior is conditional on the hazard, prior and likelihood. A constant hazard implies a geometric prior on regime duration. This online result is not equivalent to PELT's retrospective penalised segmentation.
