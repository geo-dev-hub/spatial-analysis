# Forecast verification lab
Work through binary-event verification from the contingency table to probability calibration and discrimination. The values are synthetic and deliberately small enough to check by hand.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'implementations/python/forecast-verification'))
from forecast_verification import contingency_table, categorical_scores, brier_score, brier_skill_score, reliability_bins, roc_auc
observed = [1, 1, 0, 0, 1, 0, 1, 0]
deterministic = [1, 0, 0, 1, 1, 0, 0, 0]
probabilities = [0.85, 0.60, 0.20, 0.55, 0.75, 0.15, 0.40, 0.10]
table = contingency_table(observed, deterministic)
table, categorical_scores(table)
```

## Categorical scores answer different questions
POD conditions on observed events. FAR conditions on forecast events. The false-positive rate conditions on observed non-events. CSI and ETS combine different parts of the contingency table. Do not report a score without its underlying counts.


```python
reference = [sum(observed) / len(observed)] * len(observed)
print('Brier score:', brier_score(observed, probabilities))
print('Brier skill versus sample climatology reference:', brier_skill_score(observed, probabilities, reference))
print('ROC AUC:', roc_auc(observed, probabilities))
```

## Reliability bins
Each bin compares the average forecast probability with the observed event frequency. Empty bins remain in the output; a bin with no cases contains no evidence about calibration.


```python
reliability_bins(observed, probabilities, [0, 0.25, 0.5, 0.75, 1])
```
