# Line clipping lab
Compare three segment-clipping algorithms and inspect where their input contracts differ.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'implementations/python/line-clipping'))
from line_clipping import cohen_sutherland, liang_barsky, cyrus_beck
window = (0, 0, 10, 8)
segment = (-3, 2, 12, 6)
p0, p1 = segment[:2], segment[2:]
print('Cohen–Sutherland:', cohen_sutherland(p0, p1, window))
print('Liang–Barsky:', liang_barsky(p0, p1, window))
```

## Convex polygon window
Cyrus–Beck accepts a convex polygon rather than an axis-aligned rectangle. Reverse the ring to check that orientation is handled.


```python
window_polygon = [(1, 1), (9, 1), (8, 7), (2, 7)]
print(cyrus_beck(p0, p1, window_polygon))
print(cyrus_beck(p0, p1, list(reversed(window_polygon))))
```
