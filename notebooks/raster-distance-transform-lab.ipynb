# Raster distance transform lab
Calculate centre-to-centre Euclidean distance from every raster cell to the nearest target cell. Then change the cell dimensions and inspect the result.

This is not a cost-distance or least-cost path algorithm.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'implementations').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'implementations/python/raster-distance'))
from raster_distance import distance_to_targets
```

## 1. Mark the target cells

True cells are the locations from which distance is measured. Multiple targets are allowed; each cell receives the distance to the nearest one.

```python
targets = np.zeros((11, 13), dtype=bool)
targets[5, 6] = True
targets[2, 10] = True
targets[8, 2] = True
fig, ax = plt.subplots(figsize=(7, 5))
ax.imshow(targets, cmap='Greys', interpolation='nearest')
ax.set_title('Target cells (white)'); ax.set_xlabel('Column'); ax.set_ylabel('Row'); ax.set_xticks(range(13)); ax.set_yticks(range(11)); ax.grid(True, alpha=.2)
plt.show()
```

## 2. Square cells

With unit spacing in both directions, the transform gives Euclidean distances in grid units.

```python
square = distance_to_targets(targets, cell_size=(1, 1))
fig, ax = plt.subplots(figsize=(8, 5))
im = ax.imshow(square, cmap='viridis', interpolation='nearest')
ax.contour(square, levels=[1, 2, 3, 4, 5], colors='white', linewidths=.6)
ax.scatter([6, 10, 2], [5, 2, 8], marker='x', color='red', s=70)
ax.set_title('Euclidean distance, square cells'); ax.set_xlabel('Column'); ax.set_ylabel('Row'); fig.colorbar(im, ax=ax, label='Grid units')
plt.show()
```

## 3. Rectangular cells

Now make each row step 10 metres and each column step 20 metres. The same array-index offset no longer represents the same map distance in both directions.

```python
rectangular = distance_to_targets(targets, cell_size=(10, 20))
fig, ax = plt.subplots(figsize=(8, 5))
im = ax.imshow(rectangular, cmap='magma', interpolation='nearest')
ax.set_title('Euclidean distance, 10 m × 20 m cells'); ax.set_xlabel('Column'); ax.set_ylabel('Row'); fig.colorbar(im, ax=ax, label='Metres')
plt.show()
print('One row away:', rectangular[6, 6], 'm')
print('One column away:', rectangular[5, 7], 'm')
print('Diagonal neighbour:', rectangular[6, 7], 'm')
```
