# Differential privacy composition lab
A small in-memory accountant applying basic sequential composition. It tracks declared mechanism guarantees; it does not prove that a mechanism satisfies them.

```python
import sys
sys.path.insert(0, '../implementations/python/spatial-privacy')
from composition import SequentialPrivacyAccountant
accountant = SequentialPrivacyAccountant(epsilon_budget=1.0, delta_budget=1e-5)
accountant.add_release('monthly-cell-counts', epsilon=0.25, delta=0)
accountant.add_release('district-summary', epsilon=0.30, delta=2e-6)
accountant.summary()
```

```python
print('Composed epsilon:', accountant.composed_epsilon)
print('Remaining epsilon:', accountant.remaining_epsilon)
print('Composed delta:', accountant.composed_delta)
print('Remaining delta:', accountant.remaining_delta)
try:
    accountant.add_release('third-release', epsilon=0.6, delta=1e-6)
except ValueError as error:
    print('Rejected without changing the ledger:', error)
accountant.summary()
```

## Interpretation limits
Basic sequential composition adds epsilon and delta across releases. This lab does not implement advanced composition, Rényi DP, subsampling amplification, adaptive composition refinements, durable storage or central release governance. The accountant is only as sound as the mechanism guarantees supplied to it.
