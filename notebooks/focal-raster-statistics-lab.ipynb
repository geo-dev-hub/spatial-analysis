# Focal raster statistics with integral images

A focal mean summarises a moving neighbourhood around each cell. This lab compares the raster, focal mean and valid-observation count.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'implementations').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'implementations/python/focal-statistics'))
from focal_statistics import focal_mean
```

## 1. Compare a raster with its focal mean

The window is a centred 3×3 square. At edges it is truncated to the cells that exist.

```python
raster = np.array([
    [2, 3, 5, 7, 8],
    [3, 4, 6, 8, 9],
    [4, 5, 9, 9, 10],
    [5, 7, 8, 10, 12],
    [6, 8, 9, 11, 13],
] , dtype=float)
means, counts = focal_mean(raster, window_size=3)
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, data, title in zip(axes, [raster, means, counts], ['Original values', 'Focal mean', 'Valid count']):
    im = ax.imshow(data, cmap='viridis'); ax.set_title(title)
    ax.set_xticks(range(raster.shape[1])); ax.set_yticks(range(raster.shape[0]))
    fig.colorbar(im, ax=ax)
plt.tight_layout(); plt.show()
```

## 2. NoData changes the count, not the value to zero

Invalid values are ignored. The count raster lets you see how much evidence supports each focal mean.

```python
masked = raster.copy()
masked[1, 1] = np.nan
masked[2, 2] = np.nan
means_masked, counts_masked = focal_mean(masked, window_size=3, min_count=4)
fig, axes = plt.subplots(1, 2, figsize=(9, 4))
for ax, data, title in zip(axes, [means_masked, counts_masked], ['Mean (minimum 4 valid cells)', 'Valid observations']):
    im = ax.imshow(np.ma.masked_invalid(data), cmap='viridis'); ax.set_title(title); fig.colorbar(im, ax=ax)
plt.tight_layout(); plt.show()
```

## 3. Change the neighbourhood size

A larger window smooths more broadly. Compare the same raster under different window sizes.

```python
for size in [1, 3, 5]:
    result, _ = focal_mean(raster, window_size=size)
    print(f'window={size}, centre mean={result[2, 2]:.2f}')
```
