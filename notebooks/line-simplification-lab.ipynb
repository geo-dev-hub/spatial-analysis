# Line simplification laboratory
Compare vertex retention and geometric decisions for four open-line simplifiers. Inputs are planar coordinates; the tolerance is in coordinate units except for Visvalingam–Whyatt's squared-unit area threshold.

```python
import sys
sys.path.insert(0, '../implementations/python/line-simplification')
from simplification import douglas_peucker, visvalingam_whyatt, reumann_witkam, lang
line = [(0,0),(1,0.05),(2,-0.03),(3,0.02),(4,0.0),(5,1.5),(6,2.0),(7,2.05)]
```

```python
results = {
    'Douglas-Peucker': douglas_peucker(line, 0.15),
    'Visvalingam-Whyatt': visvalingam_whyatt(line, 0.08),
    'Reumann-Witkam': reumann_witkam(line, 0.15),
    'Lang': lang(line, 0.15, lookahead=4),
}
{name: len(points) for name, points in results.items()}
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(9,5))
plt.plot(*zip(*line), marker='o', label='original')
for name, points in results.items():
    plt.plot(*zip(*points), marker='.', label=name)
plt.axis('equal'); plt.legend(); plt.grid(True)
```

## Interpretation
Compare not only vertex count but also where bends are retained. Reumann–Witkam is strip-driven, Lang is constrained by a finite look-ahead window, Douglas–Peucker uses global endpoint deviations within recursive subchains, and Visvalingam–Whyatt orders points by effective area. These are different criteria. Try a reversal, a tight loop, repeated vertices and a much larger coordinate scale.
