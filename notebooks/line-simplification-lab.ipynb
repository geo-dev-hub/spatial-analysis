# Line simplification laboratory
Compare vertex retention and geometric decisions for five open-line simplifiers. Distance tolerances use coordinate units except for Visvalingam–Whyatt's squared-unit area threshold.


```python
import sys
sys.path.insert(0, '../implementations/python/line-simplification')
from simplification import douglas_peucker, visvalingam_whyatt, reumann_witkam, lang, opheim
line = [(0,0),(1,0.05),(2,-0.03),(3,0.02),(4,0.0),(5,1.5),(6,2.0),(7,2.05)]
```

```python
results = {
    'Douglas-Peucker': douglas_peucker(line, 0.15),
    'Visvalingam-Whyatt': visvalingam_whyatt(line, 0.08),
    'Reumann-Witkam': reumann_witkam(line, 0.15),
    'Lang': lang(line, 0.15, lookahead=4),
    'Opheim': opheim(line, strip_half_width=0.15, min_distance=0.8, max_distance=2.5),
}
{name: len(points) for name, points in results.items()}
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(9,5))
plt.plot(*zip(*line), marker='o', label='original')
for name, points in results.items():
    plt.plot(*zip(*points), marker='.', label=name)
plt.axis('equal'); plt.legend(); plt.grid(True)
```

## Interpretation
Compare not only vertex count but also where bends are retained. Reumann–Witkam is strip-driven, Lang uses a fixed look-ahead count, and Opheim adds minimum/maximum radial search constraints to a moving corridor. Douglas–Peucker uses global endpoint deviations within recursive subchains; Visvalingam–Whyatt orders points by effective area. These are different criteria. Opheim's width/radius convention is explicit in the implementation. Try a reversal, a tight loop, repeated vertices and a larger coordinate scale.
