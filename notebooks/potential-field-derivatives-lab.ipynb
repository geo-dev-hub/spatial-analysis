# Potential-field derivative laboratory
Create a synthetic regular-grid anomaly and calculate spectral derivatives, analytic-signal amplitude and tilt angle. FFT periodic-boundary effects are part of the experiment.

```python
import sys, numpy as np
sys.path.insert(0, '../implementations/python/potential-field')
from spectral_derivatives import spectral_derivatives, analytic_signal_amplitude, tilt_angle
y,x=np.mgrid[-32:32,-32:32]
field=np.exp(-((x-4)**2+(y+3)**2)/80.0)+0.4*np.exp(-((x+12)**2+(y-10)**2)/25.0)
gx,gy,gz=spectral_derivatives(field,1,1)
asa=analytic_signal_amplitude(field,1,1);tilt=tilt_angle(field,1,1)
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,3,figsize=(13,4))
for a,z,title in zip(ax,[field,asa,tilt],['Potential field','Analytic-signal amplitude','Tilt angle']):
    im=a.imshow(z,origin='lower');a.set_title(title);fig.colorbar(im,ax=a,shrink=.75)
plt.tight_layout()
```

## Interpretation
The analytic-signal amplitude combines horizontal and vertical derivative magnitudes. Tilt normalises the vertical derivative by total horizontal gradient and maps it to an angle. Inspect the grid edges: the FFT assumes periodicity, so edge mismatch can introduce ringing. These attributes support interpretation but do not uniquely determine source depth or geometry.
