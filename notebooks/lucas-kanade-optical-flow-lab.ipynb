# Lucas–Kanade optical flow laboratory
Estimate a known small translation in a synthetic textured image. Inspect both flow components and the valid-texture mask.

```python
import sys, numpy as np
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'implementations/python/optical-flow'))
from scipy.ndimage import shift
from lucas_kanade import lucas_kanade
y,x=np.mgrid[0:64,0:64]
first=np.sin(.25*x)+np.cos(.2*y)+.3*np.sin(.18*x+.31*y)
second=shift(first,(0,.5),order=3,mode='wrap')
flow=lucas_kanade(first,second,window_size= 9)
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,3,figsize=(12,4))
for a,z,title in zip(ax,[flow['u'],flow['v'],flow['valid']],['Horizontal flow u','Vertical flow v','Valid texture windows']):
    im=a.imshow(z,origin='lower');a.set_title(title);fig.colorbar(im,ax=a,shrink=.8)
plt.tight_layout()
```

## Interpretation
The expected horizontal displacement is +0.5 pixels and vertical displacement is zero. Low-texture or one-direction-only windows cannot determine both components and are marked invalid. This is a single-scale small-motion method; large displacements need pyramids/warping, and map-coordinate vectors need affine/CRS conversion.
