# Curve construction lab
Compare corner cutting, interpolating splines and Bezier curves. These operations construct geometric curves; they are not the same as statistically smoothing a noisy trajectory.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / 'implementations/python/curves-smoothing'))
from curves_smoothing import chaikin, catmull_rom, bezier_curve
controls = [(0, 0), (1, 2), (3, 1), (4, 3)]
print('Chaikin:', chaikin(controls, iterations=2))
print('Centripetal Catmull–Rom sample count:', len(catmull_rom(controls, 12)))
print('Bezier endpoints:', bezier_curve(controls, 5)[0], bezier_curve(controls, 5)[-1])
```

## Compare the geometric behaviour
- Chaikin moves the interior geometry toward a rounded limit curve and preserves endpoints for open lines.
- Catmull–Rom passes through the control points; centripetal parameterisation is often a useful default for unevenly spaced controls.
- Bezier curves are controlled by a control polygon; except at the endpoints, the curve generally does not pass through the interior controls.


```python
print('Uniform Catmull–Rom:', catmull_rom(controls, 8, alpha=0.0)[:4])
print('Centripetal Catmull–Rom:', catmull_rom(controls, 8, alpha=0.5)[:4])
print('Chordal Catmull–Rom:', catmull_rom(controls, 8, alpha=1.0)[:4])
print('Closed Chaikin ring:', chaikin([(0,0),(4,0),(3,3),(0,2)], closed=True, iterations=1))
```
