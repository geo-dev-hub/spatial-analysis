# Clark–Evans: comparing nearest-neighbour spacing

The Clark–Evans R ratio compares mean nearest-neighbour distance with a homogeneous Poisson reference. This lab compares synthetic clustered, regular and uniform-random patterns. It is descriptive only; no significance test or edge correction is applied.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'implementations').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'implementations/python/clark-evans'))
from clark_evans import clark_evans_index
```

## 1. Create three point patterns

All patterns use the same rectangular window. The examples are synthetic and are intended to illustrate spacing, not represent a real process.

```python
rng = np.random.default_rng(29)
random_points = rng.uniform(0, 10, size=(80, 2))
centres = np.array([[2.5, 2.5], [7.5, 2.5], [2.5, 7.5], [7.5, 7.5]])
clustered_points = np.vstack([rng.normal(loc=centre, scale=.45, size=(20, 2)) for centre in centres])
clustered_points = np.clip(clustered_points, 0, 10)
axis = np.linspace(.625, 9.375, 8)
regular_points = np.array([(x, y) for y in axis for x in axis])
patterns = {'Clustered': clustered_points, 'Uniform random': random_points, 'Regular': regular_points}
fig, axes = plt.subplots(1, 3, figsize=(12, 4), sharex=True, sharey=True)
for ax, (name, points) in zip(axes, patterns.items()):
    ax.scatter(points[:, 0], points[:, 1], s=12)
    ax.set(title=name, xlim=(0, 10), ylim=(0, 10), aspect='equal')
    ax.grid(True)
plt.tight_layout(); plt.show()
```

## 2. Calculate R

The ratio compares the observed mean nearest-neighbour distance with the expected mean under an infinite-plane homogeneous Poisson process. Finite-window edge effects are not corrected here.

```python
for name, points in patterns.items():
    result = clark_evans_index(points, bounds=(0, 10, 0, 10))
    print(f'{name:16s} R={result.ratio:.3f} | observed={result.observed_mean_distance:.3f} | expected={result.expected_mean_distance:.3f}')
```
