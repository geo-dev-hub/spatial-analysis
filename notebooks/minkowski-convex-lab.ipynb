# Convex Minkowski sum: an edge-merge visual lab

Explore how two convex polygon boundaries combine through ordered edge vectors. The implementation uses exact rational coordinates; plotting converts them to floats only for display.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'implementations').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'implementations/python/minkowski-convex'))
from minkowski_sum import minkowski_sum
import matplotlib.pyplot as plt
from fractions import Fraction
```

## 1. Predict the result

A 4-by-2 rectangle plus a 1-by-1 rectangle should produce a 5-by-3 rectangle. This is a useful check before looking at the output.

```python
P = [(0, 0), (4, 0), (4, 2), (0, 2)]
Q = [(0, 0), (1, 0), (1, 1), (0, 1)]
S = minkowski_sum(P, Q)
S
```

```python
def close_ring(ring):
    return ring + [ring[0]]

fig, ax = plt.subplots(figsize=(7, 5))
for ring, label in [(P, 'P'), (Q, 'Q'), (S, 'P ⊕ Q')]:
    xs, ys = zip(*[(float(x), float(y)) for x, y in close_ring(ring)])
    ax.plot(xs, ys, marker='o', label=label)
    for i, (x, y) in enumerate(ring):
        ax.annotate(f'{label}{i}', (float(x), float(y)), xytext=(4, 5), textcoords='offset points')
ax.set_aspect('equal', adjustable='box')
ax.grid(True)
ax.legend()
ax.set_title('Two input polygons and their Minkowski sum')
plt.show()
```

## 2. Change the shape

Try a triangle and a pentagon. Keep both rings convex. Compare the number of input edges with the number of output edges; parallel same-direction edges can merge, so the output may have fewer than the sum of the input vertex counts.

```python
triangle = [(0, 0), (3, 0), (1, 2)]
pentagon = [(0, 0), (2, 0), (3, 1), (1, 3), (-1, 1)]
combined = minkowski_sum(triangle, pentagon)
print('input edge counts:', len(triangle), '+', len(pentagon))
print('output vertices:', len(combined))
combined
```

## 3. Test the assumptions

Reverse one ring: orientation should not change the geometric result. Then try a concave ring and observe the explicit rejection. The algorithm is not a general non-convex polygon operation.

```python
assert minkowski_sum(P, Q) == minkowski_sum(list(reversed(P)), Q)
try:
    minkowski_sum([(0, 0), (3, 0), (1, 1), (3, 2), (0, 2)], Q)
except ValueError as error:
    print(type(error).__name__ + ':', error)
```
